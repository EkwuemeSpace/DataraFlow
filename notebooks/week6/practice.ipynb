### Phase 1 Practice Task: Build and Inspect an Array

**Difficulty:** Easy | **Time:** 15 minutes

**Skills used:**
- From this phase: Creating arrays (Lesson 1.2), checking dimensions (Lesson 1.3), indexing and slicing (Lesson 1.4).
- From earlier phases: None. This is your first task.

**The task:** Create a 2-D NumPy array with the following values:

Row 1: 10, 20, 30
Row 2: 40, 50, 60
Row 3: 70, 80, 90

Print the following:
1. The number of dimensions.
2. The element in the second row, third column.
3. The last element of the third row.
4. The first two elements of the first row.

**Expected output:**
```text
Dimensions: 2
Element at row 2, column 3: 60
Last element of row 3: 90
First two elements of row 1: [10 20]
```


````python
import numpy as np

array = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(f'NUmber of dimensions: {array.ndim}')
print(f'third element in the second row: {array[1, 2]}')
print(F'Last element in the third row: {array[2, 2]}')
print(f'first two element in row 1: {array[0, 0:2]}')
````


### Phase 2 Practice Task: Two-Week Temperature Report

**Difficulty:** Easy-Medium | **Time:** 25 minutes

**Skills used:**
- From this phase: Joining arrays (Lesson 2.2), shape and reshaping (Lesson 2.1), searching and sorting (Lesson 2.3), statistics (Lesson 2.4).
- From earlier phases: Creating arrays (Phase 1, Lesson 1.2), slicing with negative indexes (Phase 1, Lesson 1.4).

**The task:** You have the daily temperatures for two weeks.

Week 1: 12, 15, 14, 18, 20, 17, 16
Week 2: 19, 21, 22, 20, 18, 17, 15

Do the following in this order:
1. Create one NumPy array for each week. Store them in `week1` and `week2`.
2. Join the two arrays end to end into a single array called `all_temps`. Print it.
3. Print the shape of `all_temps`.
4. Print the mean, median, minimum, and maximum of `all_temps`.
5. Sort `all_temps` into `sorted_temps` and print it.
6. Use `np.where` to find the indexes where the temperature equals 20. Print the result.
7. Reshape `sorted_temps` into a 2-by-7 table. Print the table.
8. Print the mean of each row of that table, using `axis=1`.
9. Print the last three temperatures of `all_temps` using one negative slice.

**Expected output:**
```text
Joined: [12 15 14 18 20 17 16 19 21 22 20 18 17 15]
Shape: (14,)
Mean: 17.428571428571427
Median: 17.5
Min: 12
Max: 22
Sorted: [12 14 15 15 16 17 17 18 18 19 20 20 21 22]
Indexes of 20: (array([ 4, 10]),)
Reshaped:
[[12 14 15 15 16 17 17]
 [18 18 19 20 20 21 22]]
Row means: [15.14285714 19.71428571]
Last 3 temperatures: [18 17 15]
```


````python
import numpy as np

week1 = np.array([12,15,14,18,20,17,16])
week2 = np.array([19,21,22,20,18,17,15])

all_temps = np.concatenate((week1, week2))
print(f'joined: {all_temps}')

print(f'joined_shape: {np.shape(all_temps)}')
print(f'mean: {round(np.mean(all_temps), 1)}')
print(f'median: {np.median(all_temps)}')
print(f'minimum: {np.amin(all_temps)}')
print(f'maximum: {np.amax(all_temps)}')

sorted_temps = np.sort(all_temps)
print(f'sorted: {sorted_temps}')

index = np.where(sorted_temps == 20)
print(f'index where temperature is 20: {index}')

reshape = sorted_temps.reshape(2,7)
print(f'reshaped 2by7: \n{reshape}')

print(f'mean of each row of the table: {np.round(np.mean(reshape, axis=1)), 2}')
print(f'last three temperature: {reshape[1, 4:]} ')
````
